## **Installing a local PostgreSQL ChEMBL Database (MacOS)**

Install the Postgres App ([here](https://postgresapp.com))

```python
#Install PostgreSQL (v18 - 2026-08-15)
%%zsh
curl -JOfSL --output-dir ~/Downloads https://github.com/PostgresApp/PostgresApp/releases/download/v2.9.6/Postgres-2.9.6-18.dmg
cd ~/Downloads 
open Postgres-2.9.6-18.dmg

# Add CLI utilities (optional)
sudo mkdir -p /etc/paths.d && echo /Applications/Postgres.app/Contents/Versions/latest/bin | sudo tee /etc/paths.d/postgresapp
```



Download the latest PostgreSQL dump from the [ChEMBL Downloads page](https://chembl.gitbook.io/chembl-interface-documentation/downloads)


```python
# Download ChEMBL PostgreSQL (v37 - 2026-08-15)
%%zsh
curl -JOfSL --output-dir {DOWNLOAD_PATH} https://ftp.ebi.ac.uk/pub/databases/chembl/ChEMBLdb/latest/chembl_37_postgresql.tar.gz
cd {DOWNLOAD_PATH}
tar -xzvf chembl_37_postgresql.tar.gz
```

### **Create a new database for ChEMBL**
This process can take some time depending on your system specifications.

```python
# Create new database
%%zsh
# dropdb chembl{XX} # Remove old ChEMBL database (optional)
createdb chembl37
pg_restore --verbose --jobs=8 --no-owner --host=localhost --port=5432 --username=postgres --dbname=chembl37 ./chembl_37/chembl_37_postgresql/chembl_37_postgresql.dmp
```
