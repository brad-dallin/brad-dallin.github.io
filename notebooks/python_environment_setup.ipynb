# **Python Environment Setup**

Below is a step-by-step guide to recreating my Python Mamba environment. The `adbc_driver_postgresql` module is used for database connectivity and common Python packages such as, `pandas` and `numpy` for data manipulation are installed with `rdkit`.

```python
# Install Micromamba using Brew
brew install micromamba

# Create RDKit mamba environment
mamba create -n rdkit-env conda-forge::rdkit

# Mamba install ADBC Driver PostgreSQL
mamba install -n rdkit-env conda-forge::adbc-driver-postgresql

# Mamba install PyArrow
mamba install -n rdkit-env conda-forge::pyarrow

# Mamba install python-dotenv
mamba install -n rdkit-env conda-forge::python-dotenv
```

The `psycopg2` module can also be used, if you prefer. Although, the `adbc_driver_postgresql` module provides a more modern interface compared to `psycopg2` and integrates well with `pandas` for data analysis workflow, see [here](https://pandas.pydata.org/docs/reference/api/pandas.read_sql.html).
