## **Python Environment Setup Using Pixi (MacOS)**

```python
# Install pixi using Brew
brew install pixi

# Pixi global install direnv
pixi global install direnv

# Add to .zshrc
echo '''# Add Pixi to PATH
export PATH=$PATH:${HOME}/.pixi/bin

# Add direnv
eval "$(direnv hook zsh)"
''' > ~/.zshrc

# Pixi create workspace
mkdir {WORKSPACE}
cd {WORKSPACE}
pixi init .

# Create .envrc file in {WORKSPACE}
echo '''watch_file pixi.lock 
eval "$(pixi shell-hook)"
''' > .envrc

# Pixi add dependencies
pixi add python
pixi add pip
pixi add ipykernel
pixi add ipywidgets
pixi add rdkit
pixi add adbc-driver-postgresql
pixi add pyarrow
pixi add python-dotenv
```
